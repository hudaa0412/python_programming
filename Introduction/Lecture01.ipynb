```python
num1 = float(input("Enter the first number :"))
num2 = float(input("Enter the second number : "))
choice = input('Enter choice(+,-,*,/)')


def sum(num1,num2):
    return(num1+num2)

def sub(num1,num2):
    return(num1-num2)

def multiply(num1,num2):
    return(num1*num2)

def divide(num1,num2):
    return(num1/num2)

if choice =='/':
    ans=divide(num1,num2)
    print(ans)


elif choice =='*':
    ans = multiply(num1,num2)
    print(ans)

elif choice =="-":
    ans = sub(num1,num2)
    print(ans)


elif choice == "+":
    ans = sum(num1,num2)
    print(ans)



```

```python
# Pipeline Concept : Orchestrator 
def sum (num1,num2):
    return (num1+num2)
def sub (num1,num2):
    return (num1-num2)
def mul (num1,num2):
    return (num1*num2)
def div (num1,num2):
    return (num1/num2)

def main():   # Fuction Definition

    num1 = float(input("Enter the first number : "))
    num2 = float(input("Enter the second number : "))
    choice = input('Enter the Choice : ')
    # Addition 
    if choice == "+":

        ans=sum(num1,num2)
    # Substraction
    elif choice == "-":
        ans=sub(num1,num2)
    # Multiplication
    elif choice == "*":
        ans=mul(num1,num2)
    # Division
    elif choice == "/":
        ans=div(num1,num2)

    else:
        print("Invalid Choice")
    
    print(ans)
 

main() # Function calling : Orschestrator
```
