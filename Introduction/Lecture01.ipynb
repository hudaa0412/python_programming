```python
train_name = "abc"

booking_available = 16

print("Train Name:", train_name)
print("Booking Available:", booking_available)

booking = input("Do you want to book a ticket? (yes/no): ")

if booking == "yes":
    tickets = int(input("How many tickets do you want? "))

    if tickets <= booking_available:
        booking_available = booking_available - tickets

        print("Booking Status: Booking Done")
        print("Number of Bookings:", tickets)
        print("Available Booking:", booking_available)

    else:
        print("Booking Status: Booking Not Done")
        print("Sorry, tickets are not available.")

else:
    print("Booking Status: Booking Not Done")
```
