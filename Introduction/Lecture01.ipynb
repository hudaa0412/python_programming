```python
# Pipeline Concept : Orchestrator 
def sum (num1,num2):
    return (num1+num2)
def sub (num1,num2):
    return (num1-num2)
def mul (num1,num2):
    return (num1*num2)
def div (num1,num2):
    return (num1/num2)

def main():   # Fuction Definition

    num1 = float(input("Enter the first number : "))
    num2 = float(input("Enter the second number : "))
    choice = input('Enter the Choice : ')
    # Addition 
    if choice == "+":

        ans=sum(num1,num2)
    # Substraction
    elif choice == "-":
        ans=sub(num1,num2)
    # Multiplication
    elif choice == "*":
        ans=mul(num1,num2)
    # Division
    elif choice == "/":
        ans=div(num1,num2)

    else:
        print("Invalid Choice")
    
    print(ans)
 

main() # Function calling : Orschestrator
```

```python
# Parent Class
class bank:

    def __init__(self, bank_name, address, ifsc_code):
        self.bank_name = bank_name
        self.address = address
        self.ifsc_code = ifsc_code

    def open_account(self):
        print("Account opened successfully.")

    def close_account(self):
        print("Account closed successfully.")

    def get_branch_info(self):
        print("Bank:", self.bank_name)
        print("Address:", self.address)
        print("IFSC Code:", self.ifsc_code)


# Child Class 1
class SavingsAccount(bank):

    def __init__(self, bank_name, address, ifsc_code,
                 interest_rate, min_balance):
        super().__init__(bank_name, address, ifsc_code)
        self.interest_rate = interest_rate
        self.min_balance = min_balance

    def credit(self, amount):
        print("Savings Account - Amount credited:", amount)

    def debit(self, amount):
        print("Savings Account - Amount debited:", amount)

    def calculate_interest(self):
        print("Interest Rate:", self.interest_rate)


# Child Class 2
class CurrentAccount(Bank):

    def __init__(self, bank_name, address, ifsc_code,
                 overdraft_limit, min_balance):
        super().__init__(bank_name, address, ifsc_code)
        self.overdraft_limit = overdraft_limit
        self.min_balance = min_balance

    def credit(self, amount):
        print("Current Account - Amount credited:", amount)

    def debit(self, amount):
        print("Current Account - Amount debited:", amount)

    def check_overdraft(self):
        print("Overdraft Limit:", self.overdraft_limit)


# Child Class 3
class FixedDeposit(Bank):

    def __init__(self, bank_name, address, ifsc_code,
                 tenure, interest_rate):
        super().__init__(bank_name, address, ifsc_code)
        self.tenure = tenure
        self.interest_rate = interest_rate

    def create_fd(self):
        print("Fixed Deposit created.")

    def calculate_maturity(self):
        print("Maturity calculated.")

    def premature_close(self):
        print("Fixed Deposit closed prematurely.")


# Child Class 4
class RecurringDeposit(Bank):

    def __init__(self, bank_name, address, ifsc_code,
                 installment_amount, tenure, interest_rate):
        super().__init__(bank_name, address, ifsc_code)
        self.installment_amount = installment_amount
        self.tenure = tenure
        self.interest_rate = interest_rate

    def create_rd(self):
        print("Recurring Deposit created.")

    def calculate_maturity(self):
        print("Maturity calculated.")

    def close_rd(self):
        print("Recurring Deposit closed.")


# Second-level Child Class
class RegularSavings(SavingsAccount):

    def __init__(self, bank_name, address, ifsc_code,
                 interest_rate, min_balance, limit_per_day):
        super().__init__(
            bank_name, address, ifsc_code,
            interest_rate, min_balance
        )
        self.limit_per_day = limit_per_day

    def credit(self, amount):
        print("Regular Savings - Amount credited:", amount)

    def debit(self, amount):
        print("Regular Savings - Amount debited:", amount)


# Second-level Child Class
class SeniorCitizenSavings(SavingsAccount):

    def __init__(self, bank_name, address, ifsc_code,
                 interest_rate, min_balance, age_limit):
        super().__init__(
            bank_name, address, ifsc_code,
            interest_rate, min_balance
        )
        self.age_limit = age_limit

    def credit(self, amount):
        print("Senior Citizen Savings - Amount credited:", amount)

    def debit(self, amount):
        print("Senior Citizen Savings - Amount debited:", amount)


# Second-level Child Class
class IndividualCurrent(CurrentAccount):

    def __init__(self, bank_name, address, ifsc_code,
                 overdraft_limit, min_balance,
                 holder_name, transaction_limit):
        super().__init__(
            bank_name, address, ifsc_code,
            overdraft_limit, min_balance
        )
        self.holder_name = holder_name
        self.transaction_limit = transaction_limit

    def credit(self, amount):
        print("Individual Current - Amount credited:", amount)

    def debit(self, amount):
        print("Individual Current - Amount debited:", amount)


# Second-level Child Class
class BusinessCurrent(CurrentAccount):

    def __init__(self, bank_name, address, ifsc_code,
                 overdraft_limit, min_balance,
                 business_name, credit_limit):
        super().__init__(
            bank_name, address, ifsc_code,
            overdraft_limit, min_balance
        )
        self.business_name = business_name
        self.credit_limit = credit_limit

    def credit(self, amount):
        print("Business Current - Amount credited:", amount)

    def debit(self, amount):
        print("Business Current - Amount debited:", amount)



# Creating an object of RegularSavings
account1 = RegularSavings(
    "HDFC Bank",
    "Kalyan",
    "HDFC0001234",
    4.5,
    5000,
    100000
)

account1.get_branch_info()
account1.credit(10000)
account1.debit(2000)
account1.calculate_interest()
```
