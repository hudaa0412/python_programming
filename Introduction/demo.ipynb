```python
# Import Important libraries
import numpy as np
import pandas as pd
# For visualization
import matplotlib.pyplot as plt
import seaborn as sns
#logging library
import logging
logging.basicConfig(level= logging.INFO,
                    filename= 'model.log',
                    filemode= 'w',
                    format="%(asctime)s-%(message)s - %(levelname)s",
                    force=True)
#filter warning messages
import warnings
warnings.filterwarnings(action = 'ignore')
#set theme
sns.set_theme(style='darkgrid')
```

```python
#step1 :Data Injestion (Loading the data)
# Pamdas is wiedly used for manipulation
filepath = "C:\python_programming\Titanic-Dataset.csv"
df = pd.read_csv(filepath)
df
```

```python
def data_injestion():

  try:
    filepath ="C:\python_programming\Titanic-Dataset.csv"
    df = pd.read_csv(filepath)
    df.sample(frac = 0.2) #shuffel the data
  except FileNotFoundError as e:
    logging.error(e)
  except Exception as e:
    logging.error(e)
  else:
    logging.info('Data Injestion Successful')
  return df
data_injestion()
```

```python
# checking Descriptive stats
numerical_col = df.select_dtypes(exclude='object').columns
categorical_col = df.select_dtypes(include = 'object').columns

from collections import OrderedDict
stats= []
for i in numerical_col:
    numerical_stats = OrderedDict({
        'Feature':i,
        'Maximum':df[i].max(),
        'Minimum' :df[i].min(),
        'Mean':df[i].mean(),
        'Median' : df[i].median(),
        'Unique_values' : df[i].nunique(),
        '% of missing values':(df[i].isnull().sum()/df.shape[0])*100,
        'Null Values': df[i].isnull().sum(),
        'Q1':df[i].quantile(0.25,),
        'Q3':df[i].quantile(0.75),
        'IQR':df[i].quantile(0.75)-df[i].quantile(0.25),
        'std':df[i].std,
        'Skewness':df[i].skew(),
        'Kurtosis':df[i].kurt()
    })
    stats.append(numerical_stats)
    report = pd.DataFrame(stats)
report
```

```python
#saving report to csv
report.to_csv('report.csv')
```
